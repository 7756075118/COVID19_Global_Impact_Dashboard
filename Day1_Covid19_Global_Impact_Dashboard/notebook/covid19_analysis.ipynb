```python

import pandas as pd
from pathlib import Path

BASE_DIR = Path(
    r"D:\Veda Internship\Sakshi\Super_Store_DataCleaning\COVID19_Global_Impact_Dashboard"
)

DATA_PATH = (
    BASE_DIR
    / "Day1_Covid19_Global_Impact_Dashboard"
    / "Dataset"
    / "owid-covid-data.csv"
)

print("Dataset path:", DATA_PATH)
print("File exists:", DATA_PATH.exists())

if DATA_PATH.exists():
    df = pd.read_csv(DATA_PATH, low_memory=False)
    print("Dataset loaded successfully!")
    print("Rows and columns:", df.shape)
    display(df.head())
else:
    print("CSV file not found. Please check the folder location.")
```

```python

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()

print("\nMissing values in important columns:")
important_cols = [
    "location", "date", "new_cases",
    "new_deaths", "total_cases", "total_deaths",
    "population", "people_vaccinated_per_hundred"
]

display(df[important_cols].isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDate range:")
print(df["date"].min(), "to", df["date"].max())

print("\nUnique locations:", df["location"].nunique())
```

```python

# Create a working copy
covid = df.copy()

# Convert date column
covid["date"] = pd.to_datetime(covid["date"], errors="coerce")

# Remove rows with missing location or invalid date
covid = covid.dropna(subset=["location", "date"])

# Keep country-level records and exclude aggregate regions
if "iso_code" in covid.columns:
    covid = covid[
        ~covid["iso_code"].astype(str).str.startswith("OWID_")
    ].copy()

# Convert analysis columns to numeric
numeric_cols = [
    "new_cases", "new_deaths",
    "total_cases", "total_deaths",
    "population", "people_vaccinated_per_hundred"
]

for col in numeric_cols:
    if col in covid.columns:
        covid[col] = pd.to_numeric(covid[col], errors="coerce")

# Sort by country and date
covid = covid.sort_values(["location", "date"])

print("Original rows:", len(df))
print("Rows after cleaning:", len(covid))
print("Remaining duplicate rows:", covid.duplicated().sum())
print("Cleaning completed!")
```

```python
import numpy as np
# Seven-day rolling averages for each country
covid["cases_7day_avg"] = (
    covid.groupby("location")["new_cases"]
    .transform(lambda s: s.rolling(7, min_periods=7).mean())
)

covid["deaths_7day_avg"] = (
    covid.groupby("location")["new_deaths"]
    .transform(lambda s: s.rolling(7, min_periods=7).mean())
)

# Population-normalized metrics
covid["cases_per_million"] = (
    covid["total_cases"] / covid["population"] * 1_000_000
)

covid["deaths_per_million"] = (
    covid["total_deaths"] / covid["population"] * 1_000_000
)

# Descriptive case fatality ratio
covid["case_fatality_rate"] = (
    covid["total_deaths"]
    / covid["total_cases"].replace(0, np.nan)
    * 100
)

display(
    covid[
        [
            "location", "date", "new_cases",
            "cases_7day_avg", "new_deaths",
            "deaths_7day_avg", "cases_per_million",
            "deaths_per_million", "case_fatality_rate"
        ]
    ].tail(10)
)
```

```python

CLEAN_DIR = BASE_DIR /"Day1_Covid19_Global_Impact_Dashboard"/ "cleaned_data"
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

cleaned_path = CLEAN_DIR / "covid19_cleaned.csv"

covid.to_csv(cleaned_path, index=False)

print("Cleaned dataset saved successfully!")
print("File location:", cleaned_path)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
CHART_DIR = BASE_DIR /"Day1_Covid19_Global_Impact_Dashboard"/ "charts"
CHART_DIR.mkdir(parents=True, exist_ok=True)

sns.set_theme(style="whitegrid")

# Chart 1: Global daily cases, seven-day average
global_cases = (
    covid.groupby("date")["new_cases"]
    .sum(min_count=1)
    .rolling(7, min_periods=7)
    .mean()
)

plt.figure(figsize=(12, 5))
global_cases.plot()
plt.title("Global COVID-19 Cases - 7-Day Average")
plt.xlabel("Date")
plt.ylabel("Average Daily Cases")
plt.tight_layout()
plt.savefig(CHART_DIR / "01_cases_trend.png", dpi=150)
plt.show()
plt.close()

# Chart 2: Global daily deaths, seven-day average
global_deaths = (
    covid.groupby("date")["new_deaths"]
    .sum(min_count=1)
    .rolling(7, min_periods=7)
    .mean()
)

plt.figure(figsize=(12, 5))
global_deaths.plot()
plt.title("Global COVID-19 Deaths - 7-Day Average")
plt.xlabel("Date")
plt.ylabel("Average Daily Deaths")
plt.tight_layout()
plt.savefig(CHART_DIR / "02_deaths_trend.png", dpi=150)
plt.show()
plt.close()

# Latest available record for each country
latest = (
    covid.dropna(subset=["cases_per_million"])
    .sort_values("date")
    .groupby("location")
    .tail(1)
)

# Chart 3: Top 10 countries by reported cases per million
top10 = latest.nlargest(10, "cases_per_million")

plt.figure(figsize=(11, 6))
sns.barplot(data=top10, x="cases_per_million", y="location")
plt.title("Top 10 Countries by Reported Cases per Million")
plt.xlabel("Reported Cases per Million")
plt.ylabel("Country")
plt.tight_layout()
plt.savefig(CHART_DIR / "03_country_comparison.png", dpi=150)
plt.show()
plt.close()

# Chart 4: Vaccination progress
countries = ["India", "United States", "United Kingdom", "Brazil"]

vacc = covid[covid["location"].isin(countries)].copy()
vacc = vacc.dropna(subset=["people_vaccinated_per_hundred"])

if not vacc.empty:
    plt.figure(figsize=(12, 6))
    sns.lineplot(
        data=vacc,
        x="date",
        y="people_vaccinated_per_hundred",
        hue="location"
    )
    plt.title("COVID-19 Vaccination Progress")
    plt.xlabel("Date")
    plt.ylabel("People Vaccinated per 100 People")
    plt.tight_layout()
    plt.savefig(CHART_DIR / "04_vaccination_trend.png", dpi=150)
    plt.show()
    plt.close()

print("Charts saved in:", CHART_DIR)
```
